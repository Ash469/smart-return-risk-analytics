# Initial Data Exploration (EDA)
This notebook loads the raw Olist CSVs and performs Exploratory Data Analysis (EDA) to find patterns that justify our Machine Learning approach.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')
```

```python
raw_dir = '../data/raw/'
orders = pd.read_csv(os.path.join(raw_dir, 'olist_orders_dataset.csv'))
customers = pd.read_csv(os.path.join(raw_dir, 'olist_customers_dataset.csv'))
products = pd.read_csv(os.path.join(raw_dir, 'olist_products_dataset.csv'))
order_items = pd.read_csv(os.path.join(raw_dir, 'olist_order_items_dataset.csv'))
reviews = pd.read_csv(os.path.join(raw_dir, 'olist_order_reviews_dataset.csv'))
```

### Geographic Distribution of Customers

```python
plt.figure(figsize=(14, 6))
state_counts = customers['customer_state'].value_counts()
sns.barplot(x=state_counts.index, y=state_counts.values, palette='coolwarm')
plt.title('Customer Geographic Distribution by State', fontsize=16, fontweight='bold')
plt.xlabel('State Code', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.show()
```

### Orders Over Time

```python
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['purchase_month_year'] = orders['order_purchase_timestamp'].dt.to_period('M')
monthly_orders = orders.groupby('purchase_month_year').size().reset_index(name='order_count')
monthly_orders['purchase_month_year'] = monthly_orders['purchase_month_year'].astype(str)

plt.figure(figsize=(16, 6))
sns.lineplot(data=monthly_orders, x='purchase_month_year', y='order_count', marker='o', color='b', linewidth=2.5)
plt.xticks(rotation=45)
plt.title('Order Volume Over Time (2016-2018)', fontsize=16, fontweight='bold')
plt.xlabel('Month-Year', fontsize=12)
plt.ylabel('Total Orders', fontsize=12)
plt.show()
```

### Review Score Distribution

```python
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=reviews, x='review_score', palette='RdYlGn')
plt.title('Distribution of Customer Review Scores', fontsize=16, fontweight='bold')
plt.xlabel('Review Score (1 to 5)', fontsize=12)
plt.ylabel('Number of Reviews', fontsize=12)
plt.show()
```

 **Conclusion:** 1-star and 2-star reviews indicate deep buyer dissatisfaction. We will use this pool of low-score reviews to anchor our return simulation, assuming that highly dissatisfied customers are the most likely to return items.

### Price & Freight Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
sns.histplot(order_items['price'], bins=50, kde=True, ax=axes[0], color='purple')
axes[0].set_xlim(0, 500)
axes[0].set_title('Distribution of Item Prices (Zoomed 0-$500)', fontsize=14)
axes[0].set_xlabel('Price ($)')

sns.histplot(order_items['freight_value'], bins=50, kde=True, ax=axes[1], color='orange')
axes[1].set_xlim(0, 100)
axes[1].set_title('Distribution of Freight Costs (Zoomed 0-$100)', fontsize=14)
axes[1].set_xlabel('Freight Cost ($)')
plt.show()
```

Both distributions are highly right-skewed. Most items cost between $20 and $150, and most shipping costs are between $10 and $30. 
- **Conclusion:** Because shipping costs are relatively high compared to item costs, the ratio between the two is likely a critical psychological factor for buyers.That is why i have created `freight_to_value_ratio` feature using SQL.

### Order Status Distribution

```python
plt.figure(figsize=(10, 5))
ax = sns.countplot(data=orders, y='order_status', order=orders['order_status'].value_counts().index, palette='Blues_r')
plt.title('Distribution of Order Statuses', fontsize=16, fontweight='bold')
plt.xlabel('Number of Orders', fontsize=12)
plt.ylabel('Order Status', fontsize=12)

total = len(orders)
for p in ax.patches:
    percentage = f'{100 * p.get_width() / total:.1f}%'
    x = p.get_x() + p.get_width() + 500
    y = p.get_y() + p.get_height() / 2
    ax.annotate(percentage, (x, y), va='center')

plt.show()
```

- **Conclusion:** A product must be delivered before it can be physically returned. This explicitly validates our SQL `WHERE` clause that filtered out canceled or processing orders before calculating delivery delays.

### Top Product Categories

```python
plt.figure(figsize=(12, 6))
top_cats = products['product_category_name'].value_counts().head(10)
sns.barplot(x=top_cats.values, y=top_cats.index, palette='viridis')
plt.title('Top 10 Product Categories by Volume', fontsize=16, fontweight='bold')
plt.xlabel('Number of Unique Products', fontsize=12)
plt.ylabel('Category', fontsize=12)
plt.show()
```

The dataset features a massive variety of goods, with top categories including Bed/Bath, Sports, Furniture, and Health/Beauty.
- **Conclusion:** Since return rates vary wildly between categories (e.g., clothes are returned more often than books), we must retain category-level metrics in our Machine Learning dataset.
