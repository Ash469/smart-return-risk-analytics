# Initial Data Exploration (EDA)
In this notebook, i load the raw Olist dataset to understand its structure, identify missing values, and visualize the base distributions. This justifies  why i need to build an aggregated, order-level dataset using SQL.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid", palette="muted")
import warnings
warnings.filterwarnings('ignore')
```

### 1. Load Raw Data

```python
raw_dir = '../data/raw/'
orders = pd.read_csv(os.path.join(raw_dir, 'olist_orders_dataset.csv'))
customers = pd.read_csv(os.path.join(raw_dir, 'olist_customers_dataset.csv'))
products = pd.read_csv(os.path.join(raw_dir, 'olist_products_dataset.csv'))

print(f"Orders: {orders.shape}")
print(f"Customers: {customers.shape}")
print(f"Products: {products.shape}")
```

### 2. Order Status Distribution
Why do we only look at 'delivered' orders?

```python
plt.figure(figsize=(10, 5))
ax = sns.countplot(data=orders, y='order_status', order=orders['order_status'].value_counts().index)
plt.title('Distribution of Order Statuses', fontsize=14, fontweight='bold')
plt.xlabel('Number of Orders')
plt.ylabel('Order Status')

total = len(orders)
for p in ax.patches:
    percentage = f'{100 * p.get_width() / total:.1f}%'
    x = p.get_x() + p.get_width() + 500
    y = p.get_y() + p.get_height() / 2
    ax.annotate(percentage, (x, y), va='center')

plt.show()
# Insight: 97% of orders are delivered. We can only simulate returns on completed deliveries.
```

### 3. Top Product Categories
What are people buying?

```python
plt.figure(figsize=(12, 6))
top_cats = products['product_category_name'].value_counts().head(10)
sns.barplot(x=top_cats.values, y=top_cats.index, palette='viridis')
plt.title('Top 10 Product Categories', fontsize=14, fontweight='bold')
plt.xlabel('Number of Unique Products')
plt.ylabel('Category')
plt.show()
```
